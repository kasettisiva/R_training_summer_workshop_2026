## Where R fits in an AI workflow

A useful way to explain R's role is this:

```text
Raw data
  -> data cleaning and transformation
  -> exploratory data analysis and visualization
  -> statistical modeling / machine learning
  -> model evaluation and interpretation
  -> reports, dashboards, and communication
  -> optional AI assistance for coding, explanation, and summarization
```

R is especially strong for data preparation, statistics, visualization, model evaluation, and reproducible reporting.

## AI terms in plain language

| Term | Plain meaning | Example in R |
|---|---|---|
| Statistical learning | Use data and statistical models to understand or predict outcomes | `lm()`, `glm()` |
| Machine learning | Train models to make predictions or classifications | decision trees, random forests, clustering |
| Deep learning | Neural-network-based learning, often for large or complex data | `keras`, `torch` |
| Generative AI / LLMs | AI systems that generate text, code, summaries, and explanations | AI assistants, APIs, `ellmer` |

For an introductory R workshop, the best focus is not advanced deep learning. The best focus is learning enough R to inspect data, write code, evaluate outputs, and verify AI-generated suggestions.

## Example 1: traditional statistical learning in R

Linear regression is not new, but it is still a foundation for many predictive modeling workflows.

Here we use the built-in `mtcars` dataset to predict fuel efficiency (`mpg`) using car weight (`wt`) and horsepower (`hp`).

````r
# Inspect the dataset
head(mtcars)

# Fit a simple linear regression model
lm_model <- lm(mpg ~ wt + hp, data = mtcars)

# View model summary
summary(lm_model)
````

### Discussion questions

After running the previous cell, ask students:

* What is the outcome variable?
* What are the predictor variables?
* Does higher car weight appear to increase or decrease MPG?
* Why should we be careful before treating this as a perfect prediction model?

## Example 2: prediction using the model

Once a model is fitted, we can use it to make predictions on new data.

````r
# Create a new car example
new_car <- data.frame(wt = 3.2, hp = 140)

# Predict MPG for the new car
predict(lm_model, newdata = new_car)
````

## Example 3: train/test split for model evaluation

A common machine learning habit is to split data into training and testing sets.

* Training data: used to fit the model
* Testing data: used to evaluate how well the model works on data it did not see during training

````r
set.seed(123)

# Randomly choose 70% of rows for training
train_index <- sample(1:nrow(mtcars), size = 0.7 * nrow(mtcars))

train_data <- mtcars[train_index, ]
test_data  <- mtcars[-train_index, ]

# Fit model on training data
model_train <- lm(mpg ~ wt + hp, data = train_data)

# Predict on testing data
test_predictions <- predict(model_train, newdata = test_data)

# Compare observed and predicted values
results <- data.frame(
  observed_mpg = test_data$mpg,
  predicted_mpg = test_predictions
)

results
````

## Example 4: simple model error

A simple way to evaluate a regression model is to compute prediction error.

Here we calculate mean absolute error: the average absolute difference between observed and predicted values.

````r
results$error <- abs(results$observed_mpg - results$predicted_mpg)
mean_absolute_error <- mean(results$error)
mean_absolute_error
````

## Example 5: visualization of model results

Visualization helps us see whether predictions are close to observed values.

````r
library(ggplot2)

ggplot(results, aes(x = observed_mpg, y = predicted_mpg)) +
  geom_point(size = 3) +
  geom_abline(intercept = 0, slope = 1, linetype = "dashed") +
  labs(
    title = "Observed vs Predicted MPG",
    x = "Observed MPG",
    y = "Predicted MPG"
  ) +
  theme_minimal()
````

## How generative AI can help R learners

Generative AI tools can help students and researchers:

* explain unfamiliar R code
* debug error messages
* generate starter code
* convert base R code to `ggplot2`
* summarize model output in plain English
* suggest checks for missing values or data quality
* draft comments and documentation

However, AI-generated code should always be verified by running it, inspecting intermediate objects, and checking whether the result makes sense.

## Good prompt patterns for R help

A useful prompt usually includes:

1. context: what you are trying to do
2. the exact R code
3. the exact error message or output
4. what you already tried
5. the level of explanation you want

Example prompt:

```text
I am learning R in an introductory workshop. I used this code:

model <- lm(mpg ~ wt + hp, data = mtcars)
summary(model)

Please explain the output in plain English. Identify the outcome variable, predictors, and limitations. Do not add advanced statistics beyond an introductory level.
```

````r
# Store an AI prompt as text in R
my_prompt <- "I fit the model lm(mpg ~ wt + hp, data = mtcars) in R. Explain the outcome variable, predictors, main interpretation, and limitations in plain English for a beginner."

cat(my_prompt)
````

## Responsible AI checklist for R code

Before trusting AI-generated R code, check:

* Does the code run without errors?
* Are the package names correct?
* Are the column names correct?
* Did the code silently remove missing values?
* Are the results reasonable?
* Can you explain each major line of code?
* Did you verify the result using a small example?
* Are you following the course or project rules for AI use?

Important message for students:

> AI can help you write R code faster, but you still need R knowledge to verify, debug, and explain the result.
