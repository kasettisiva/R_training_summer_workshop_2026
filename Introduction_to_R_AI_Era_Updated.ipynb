# Introduction

## What is R?

R is a programming language and software environment for statistical computing, data analysis, visualization, and reproducible research.

R is commonly used for:

* importing, cleaning, and transforming data
* statistical analysis and modeling
* visualization and reporting
* machine learning workflows
* reproducible notebooks and documents
* connecting to modern AI tools through APIs or helper packages

## Why update R training in the AI era?

Since 2022, AI tools and large language models have become common in student, research, and workplace workflows. This does **not** make R outdated. Instead, it changes how students may use R.

A modern R user should know how to:

* understand and write basic R code independently
* use AI tools to accelerate learning and coding
* verify AI-generated code instead of trusting it blindly
* build simple predictive models
* communicate model results clearly
* recognize when a problem is data analysis, machine learning, or generative AI

## R in the AI workflow

A useful way to explain R's role is this:

```text
Raw data
  ↓
Data cleaning and transformation
  ↓
Exploratory data analysis and visualization
  ↓
Statistical modeling / machine learning
  ↓
Model evaluation and interpretation
  ↓
Reports, dashboards, and communication
  ↓
Optional: generative AI assistance for code, explanation, and summarization
```

R is especially strong in the middle of this workflow: data preparation, statistics, visualization, model evaluation, and reporting.

## Traditional statistical learning example

Linear regression is not new, but it is still a foundation for many predictive modeling workflows.

Here we predict miles per gallon using car weight and horsepower.

````r
lm_model <- lm(mpg ~ wt + hp, data = mtcars)
summary(lm_model)
````

### Interpret the model in plain language

After running the previous cell, ask students:

* Which variables are used as predictors?
* What is the outcome variable?
* Are heavier cars associated with lower or higher MPG?
* How would you explain the result to a non-technical audience?

This is a good place to show how generative AI can help explain output, but students must verify the explanation.

````r
new_cars <- data.frame(
  wt = c(2.5, 3.0, 4.0),
  hp = c(90, 120, 180)
)

predict(lm_model, newdata = new_cars)
````

## Machine learning workflow example

A simple machine learning workflow usually includes:

```text
1. Load data
2. Explore data
3. Split into training and testing data
4. Train a model
5. Predict on unseen test data
6. Evaluate performance
7. Explain limitations
```

The next example uses the built-in `iris` dataset.

````r
head(iris)
summary(iris)
table(iris$Species)
````

````r
ggplot(iris, aes(x = Petal.Length, y = Petal.Width, color = Species)) +
  geom_point(size = 2) +
  labs(title = "Iris flower measurements by species")
````

## Train/test split

````r
set.seed(123)

n <- nrow(iris)
train_index <- sample(seq_len(n), size = 0.8 * n)

iris_train <- iris[train_index, ]
iris_test  <- iris[-train_index, ]

nrow(iris_train)
nrow(iris_test)
````

## Train a classification model

We will use a decision tree because it is easy to explain visually.

````r
library(rpart)

iris_tree <- rpart(Species ~ ., data = iris_train, method = "class")
iris_tree
````

````r
library(rpart.plot)
rpart.plot(iris_tree)
````

## Predict and evaluate

````r
iris_pred <- predict(iris_tree, newdata = iris_test, type = "class")

results <- data.frame(
  actual = iris_test$Species,
  predicted = iris_pred
)

head(results)
````

````r
confusion_matrix <- table(
  Actual = results$actual,
  Predicted = results$predicted
)

confusion_matrix

accuracy <- mean(results$actual == results$predicted)
accuracy
````

## Important teaching point

A model is not useful just because it runs. We need to ask:

* Was the data appropriate?
* Was the test data separate from the training data?
* Is the accuracy meaningful?
* Are some classes harder to predict than others?
* Would the model work on new real-world data?

This is where domain knowledge and statistical thinking matter.

# Responsible Use of AI in R Programming

## Checklist before trusting AI-generated R code

Use this checklist whenever AI gives you R code:

1. **Run the code**: Does it execute without errors?
2. **Check the data columns**: Are the column names correct?
3. **Check packages**: Are the required packages installed and loaded?
4. **Check assumptions**: Is the model appropriate for the data?
5. **Check output**: Does the result make sense?
6. **Check reproducibility**: Did you set a seed if randomness is involved?
7. **Check privacy**: Do not paste sensitive or restricted data into public AI tools.
8. **Check academic policy**: Follow course or institution rules for AI use.

## AI should support learning, not replace learning

A good rule for students:

> Use AI to help you understand R code, but do not submit code you cannot explain.

For training purposes, students should be able to answer:

* What does each line do?
* What input does the code need?
* What output does the code create?
* What could go wrong?
* How do we verify the result?
